#### **IMPORT LIBRARIES**

```python
#StateGraph helps to create a Grpha and state it
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

#### **DEFINE STATE**

```python
class TemperatureState(TypedDict):
    temp_celsius: float
    temp_fahrenheit: float
    weather_label: str
```

#### **CONVERT_TEMP FUNCTION FIRST NODE 1**

```python
def convert_temp(state: TemperatureState):
    celsius = state['temp_celsius']
    # Convert Celsius to Fahrenheit
    fahrenheit = (celsius * 9/5) + 32
    state['temp_fahrenheit'] = round(fahrenheit, 2)
    return state
```

#### **LABEL_WEATHER FUNCTION NODE 2**

```python
def label_weather(state: TemperatureState):
    fahrenheit = state['temp_fahrenheit']

    if fahrenheit >= 86:
        state['weather_label'] = "HOT"
    elif fahrenheit >= 68:
        state['weather_label'] = "WARM"
    else:
        state['weather_label'] = "COLD"

    return state
```

#### **DEFINE AND COMPILE GRAPH**

```python
# Define your Graph
graph = StateGraph(TemperatureState)

# Add the nodes to the graph
graph.add_node('convert_temp', convert_temp)
graph.add_node('label_weather', label_weather)

# Add edges to the graph
graph.add_edge(START, 'convert_temp')
graph.add_edge('convert_temp', 'label_weather')
graph.add_edge('label_weather', END)

# Compile the graph
workflow = graph.compile()
```

#### **EXECUTE THE GRAPH**

```python
# Execute the graph
initial_state = {'temp_celsius': 28.5}
final_state = workflow.invoke(initial_state)
print(final_state)
```
