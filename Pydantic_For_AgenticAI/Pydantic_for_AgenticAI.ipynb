# **What is Pydantic**

[Pydantic](https://docs.pydantic.dev/) is a powerful data validation and parsing library for Python. It uses Python's native type hints to guarantee that your data structures conform to the types you declare at runtime.

### **Why use Pydantic?**
- **Type Safety:** Automatically validates input types.
- **Data Parsing:** Coerces input values to the correct type (e.g., parsing the string `'123'` into the integer `123`).
- **Detailed Errors:** Returns clear error messages explaining exactly which fields failed validation and why.
- **Performance:** Its core validation engine is written in Rust, making it extremely fast.

## **Key Features & Benefits of Pydantic**

1. **Type Coercion (Data Parsing):** Pydantic doesn't just validate types; it actively parses input values into the declared types wherever possible (e.g., converting strings to integers, floats, or booleans).
2. **Custom Validation Rules:** You can easily write custom validators to enforce complex rules on individual fields or across multiple fields.
3. **Settings Management:** Pydantic is widely used to load and validate application configurations or environment variables securely.
4. **Serialization:** Easily convert models back to Python dictionaries, JSON strings, or auto-generate JSON schemas.
5. **Seamless Ecosystem Integration:** Powering libraries like **FastAPI**, SQLModel, and LangChain, it is the standard for modern Python data pipelines.

## **The Problem Without Pydantic**

Without Pydantic, if you receive raw data (e.g., from an API or a JSON file), you have to manually write extensive code to:
1. Check if all required fields are present.
2. Manually cast types (like converting string `'29.99'` to float `29.99`).
3. Write conditional statements to validate rules (like ensuring age > 0).
4. Capture and collect all errors manually so you can report them to the client.

## **How to Use Pydantic**

To use Pydantic, you define the structure of your data by creating a class that inherits from `pydantic.BaseModel`. You then use Python type hints to specify the expected types of your fields.



```python
from pydantic import BaseModel

class PatientData(BaseModel):
  name: str
  age: int
  weight: int

def add_patient_data(patient : PatientData):
  print(patient.name)
  print(patient.age)
  print(patient.weight)
  print()
  print("Data Added Successfully to the database!")

def update_patient_data(patient : PatientData):
  print(patient.name)
  print(patient.age)
  print(patient.weight)
  print()
  print("Data Updated Successfully to the database!")

# Corrected the key from 'Weight' to 'weight' to match the schema exactly
patient_data = {"name": "Akhil", "age": 20, "weight": 53}
patient_1 = PatientData(**patient_data)
add_patient_data(patient_1)
#update_patient_data(patient)
```

### **Required and Optional Fields**

```python
from pydantic import BaseModel
from typing import Optional

class UserProfile(BaseModel):
    # Required field
    username: str

    # Optional field (defaults to None if not provided)
    email: Optional[str] = None

    # Field with a default value
    role: str = "patient"

# Scenario 1: Providing only the required fields
user_1 = UserProfile(username="akhil_99")
print("User 1 (Only required fields):", user_1)

# Scenario 2: Providing all fields
user_2 = UserProfile(username="aryan_dev", email="aryan@example.com", role="admin")
print("User 2 (All fields specified):", user_2)
```

### **Field Validator**

```python
#!pip install -q email-validator

from pydantic import BaseModel, EmailStr, AnyUrl, Field, field_validator, ValidationError
from typing import List, Dict, Optional, Annotated

class PatientData(BaseModel):
  name: Annotated[str,  Field(min_length=3, max_length=20, title="Name of the patient", description="Give the name of the patient")]
  age: int = Field(gt=0, lt=120)
  email: EmailStr
  linkeding_url: AnyUrl
  weight: int
  married: bool
  allergies: list[str] = Field(min_length=1)
  contact_info: dict[str, str]

  @field_validator('email')
  @classmethod
  def email_validator(cls, value):
    valid_domains = ['hdfc.com', 'icici.com']

    domain_name = value.split('@')[-1]

    if domain_name not in valid_domains:
      raise ValueError("Invalid email domain. Allowed domains: hdfc.com, icici.com")

    return value

  @field_validator('name')
  @classmethod

  def name_validator(cld, value):
    return value.upper

def add_patient_data(patient : PatientData):
  print(f"Name: {patient.name}")
  print(f"Age: {patient.age}")
  print(f"Email: {patient.email}")
  print(f"LinkedIn: {patient.linkeding_url}")
  print(f"Weight: {patient.weight}")
  print(f"Married: {patient.married}")
  print(f"Allergies: {patient.allergies}")
  print(f"Contact Info: {patient.contact_info}")
  print()
  print("Data Added Successfully to the database!")

# Scenario 1: Demonstrating validation failure with an invalid email domain
try:
  invalid_patient_data = {
    "name": "Akhil",
    "age": 20,
    "email": "jack123@example.com",  # This domain is not in the allowed list
    "linkeding_url": "https://www.linkedin.com/in/jack",
    "weight": 53,
    "married": "True",
    "allergies": ["Peanuts", "curd"],
    "contact_info": {"Akhil": "jack123@gmail.com"}
  }
  PatientData(**invalid_patient_data)
except ValidationError as e:
  print("--- Validation Failed as expected ---")
  print(e)
  print()

# Scenario 2: Successful instantiation with a valid domain ('hdfc.com')
valid_patient_data = {
  "name": "Akhil",
  "age": 20,
  "email": "jack123@hdfc.com",  # Valid domain
  "linkeding_url": "https://www.linkedin.com/in/jack",
  "weight": 53,
  "married": "True",
  "allergies": ["Peanuts", "curd"],
  "contact_info": {"Akhil": "jack123@hdfc.com"}
}

print("--- Validation Succeeded ---")
patient_1 = PatientData(**valid_patient_data)
add_patient_data(patient_1)
```

### **Model Validator**

```python
from pydantic import BaseModel, model_validator, ValidationError
from typing import Self

class PatientData(BaseModel):
    name: str
    age: int
    married: bool

    # mode='after' means the validator runs after individual field validation
    @model_validator(mode='after')
    def check_age_and_marriage(self) -> Self:
        if self.age < 18 and self.married:
            raise ValueError("A patient under 18 cannot be marked as married.")
        return self

# This will succeed normally
valid_patient = PatientData(name="Aryan", age=21, married=True)
print(f"Success: {valid_patient.name} added.")

# Wrapped in a try-except block to handle the ValidationError and prevent crashing
try:
    invalid_patient = PatientData(name="Akhil", age=16, married=True)
except ValidationError as e:
    print("\n--- Validation Failed as expected ---")
    print(e)
```

### **Computed Feilds**

```python
from pydantic import BaseModel, computed_field

class PatientData(BaseModel):
    name: str
    weight_kg: float
    height_m: float

    # The @computed_field decorator includes this property in the model's dictionary/JSON output
    @computed_field
    def bmi(self) -> float:
        return round(self.weight_kg / (self.height_m ** 2), 2)

# Create a patient instance
patient = PatientData(name="Akhil", weight_kg=53.0, height_m=1.65)

# Access the computed field directly
print(f"Patient Name: {patient.name}")
print(f"Calculated BMI: {patient.bmi}")
print("\n--- Full Model Data ---")
print(patient.model_dump())
```
