### **SYNCHRONOUS PROGRAMMING**

````python
import time
````

````python
def fetch_weather():
  print("Fetvhing weather data...")
  time.sleep(4)
  print("Weather Data Fetched")

def fetch_news():
  print("Fetching New Data")
  time.sleep(3)
  print("News Data Fetched")


def main():
  start_time = time.time()

  fetch_weather()
  fetch_news()

  end_time = time.time()

  print(f"Total time taken: {end_time - start_time} seconds")

if __name__ == "__main__":
  main()
````

### **ASYNCHRONOUS PROGRAMMING**

````python
import asyncio
import time
````

````python
async def fetch_weather():
  print("Fetvhing weather data...")
  await asyncio.sleep(4)
  print("Weather Data Fetched")

async def fetch_news():
  print("Fetching New Data")
  await asyncio.sleep(3)
  print("News Data Fetched")


async def main():
  start_time = time.time()

  await asyncio.gather(fetch_weather(), fetch_news())

  end_time = time.time()

  print(f"Total time taken: {end_time - start_time} seconds")


await main()
````

## **Concurrency vs. Parallelism**

While they sound similar, they represent very different computing concepts:

| Concept | Analogy | Best For | Hardware Requirement |
| :--- | :--- | :--- | :--- |
| **Concurrency** (Dealing with many things at once) | **1 Barista multitasking:** Preparing espresso, steaming milk, and taking orders interchangeably. | **I/O-bound tasks** (e.g., API requests, web scraping, slow database queries). | Single-core or multi-core CPUs. |
| **Parallelism** (Doing many things at once) | **2 Baristas working together:** Two people simultaneously making drinks at separate espresso machines. | **CPU-bound tasks** (e.g., machine learning training, image processing, heavy math calculations). | Multi-core processors or multiple CPUs. |

### **How it relates to your code:**
- **Cell `CSmyGVFoGve6` (Synchronous):** The code blocks sequentially. `fetch_news()` cannot start until `fetch_weather()` finishes (Total: ~7 seconds).
- **Cell `Ea213LcBJatL` (Asynchronous / Concurrent):** Your tasks run concurrently. While `fetch_weather` pauses for its I/O delay (`await asyncio.sleep(4)`), Python yields control back to the event loop, allowing `fetch_news` to run and sleep concurrently. Because they overlap, the total time is bound only by the slowest task (Total: ~4 seconds).

## **Visualizing the Workflows**

### **1. Synchronous Execution (Sequential)**
Tasks run strictly one after the other on a single core. If a task is waiting (sleeping/blocking), the entire CPU remains idle.
```text
Core 0: [  Fetch Weather (4s)  ] [ Fetch News (3s) ]
Time:   |------------------------|-----------------|--> (7 Seconds Total)
```

---

### **2. Asynchronous Concurrency (Cooperative Multitasking)**
Tasks run on a single core but dynamically yield control when waiting. They take turns, overlapping their waiting periods.
```text
Core 0: [Fetch Weather (Start)]
        |                       [Fetch News (3s)]
        |                       |                 [Fetch Weather (End)]
Time:   |-----------------------|-----------------|--------------------> (4 Seconds Total)
```

---

### **3. True Parallelism (Multi-Core)**
Tasks physically run at the exact same time on entirely separate physical CPU cores.
```text
Core 0: [     Fetch Weather (4s)     ]
Core 1: [   Fetch News (3s)   ]
Time:   |----------------------------|--> (4 Seconds Total)
```
